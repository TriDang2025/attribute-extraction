Model: Llama-3.1-8B-Instruct, bf16, vLLM, temperature 0, guided JSON (setting `llama8b`), batch-invariant mode on, engine seed 0 (as wf_v4.9).
Runtime: **A100 (40 GB or more)**; cell 0 checks it. **One short GPU session**: model load about 4 min, then a few minutes of model time.

- **G4** (cell 8): the loop-1 columns that the G3 context split changed get a new name from the frozen Step 3 names prompt (3 calls:
  dev50 train 4 columns, dev50b train 2 + hotel 3). Prompts are in git (`results/wf_v50/g4/<set>/prompts/`).

- **G5** (cell 10): Step 4 variant **ARM** = A (pass 2 in batches of 20) + re-ask r1 (v4.9) + re-ask r2 (a batch whose r1 answer still sends half
  of an entity's columns to "unused" is asked once more with a second note) + same-entity table merge (B+M, code), seeds 0-2, dev50 and dev50b
  = 6 chains -> `results/wf_v50/g5/<set>/step4_ARM/`.
- Cells 1-6: setup, as in wf_v4.9. Cell 2 installs, then restart the session, rerun cells 0-1 and skip cell 2.
  Cell 3 (models) must run before any other cell imports `s1a`.
- Cell 11: vLLM settings + GPU session time, then **Runtime > Disconnect and delete runtime**, and tell Claude "GPU done".
- After a crash: rerun cells 0-1, 3-7 and the cell that failed (answered calls are cached per folder).
- Colab Secrets `HF_TOKEN` and `GH_TOKEN`, notebook access on. Never put a token in a cell.

```python
#@title 3. Data and the loop-2 models (MultiWOZ zip + ids from HF, GLiNER, deberta-v3-base, bge-m3; pinned revisions)
import json, os
from pathlib import Path
from huggingface_hub import hf_hub_download, snapshot_download
MODELS = Path("/content/models")
os.environ["CHK_MODELS"] = str(MODELS)          # read by chunking/chk/config.py and s1a/spans.py at import
DATA = Path("/content/attribute-extraction/chunking/data")
for f in ("MULTIWOZ2.4.zip", "conversation_ids.txt"):
    if not (DATA / f).exists():
        hf_hub_download("database-lab/attribute-extraction-data", f"chunking/{f}", repo_type="dataset", local_dir=DATA.parent)
snapshot_download("urchade/gliner_medium-v2.1", revision="40ec419335d09393f298636f471328b722c6da9e", local_dir=MODELS / "gliner_medium-v2.1")
snapshot_download("microsoft/deberta-v3-base", revision="8ccc9b6f36199bec6961081d44eb72fb3f7353f3", local_dir=MODELS / "deberta-v3-base",
                  allow_patterns=["config.json", "spm.model", "tokenizer_config.json"])
snapshot_download("BAAI/bge-m3", revision="5617a9f61b028005a4858fdac845db406aefb181", local_dir=MODELS / "bge-m3")
cfg_path = MODELS / "gliner_medium-v2.1" / "gliner_config.json"   # point GLiNER at the local backbone, as on the laptop
cfg = json.loads(cfg_path.read_text())
cfg["model_name"] = str(MODELS / "deberta-v3-base")
cfg_path.write_text(json.dumps(cfg, indent=2))
import sys
sys.path.insert(0, "/content/attribute-extraction/step1a"); sys.path.insert(0, "/content/attribute-extraction/step1a/scripts")
from s1a import spans as SP
g = SP.load()
print("GLiNER ok:", g.inference(["I need a hotel with free wifi."], ["internet access"], threshold=0.3))
del g
```

```python
#@title 4. Model setting (config/wf_v41_models.json) and GPU check
SETTING = "llama8b"  #@param ["llama8b"]
import sys
sys.path.insert(0, "/content/attribute-extraction/step1a")
from s1a import nov16_v41 as NV
SET = NV.model_setting(SETTING)
print(SETTING, SET)
!nvidia-smi --query-gpu=name,memory.used,memory.total --format=csv,noheader
import torch
assert torch.cuda.is_available(), "no GPU: change the runtime type to A100"
gb = torch.cuda.get_device_properties(0).total_memory / 2**30
print(f"GPU memory {gb:.1f} GiB")
assert gb >= SET["min_gpu_gb"], f"only {gb:.1f} GiB: {SETTING} needs {SET['min_gpu_gb']} GiB (an A100)"
```

```python
#@title 5. Load the model (gpu_memory_utilization leaves room for bge-m3)
MODEL, REVISION = SET["model"], SET["revision"]   # from the setting of cell 3
GPU_UTIL = 0.6  #@param {type:"number"}
BATCH_INVARIANT = True  #@param {type:"boolean"}
import os
os.environ["VLLM_BATCH_INVARIANT"] = "1" if BATCH_INVARIANT else "0"   # read when vllm is imported
MAX_LEN = 32768
import vllm
from vllm import LLM, SamplingParams
print("vllm", vllm.__version__, "| model", MODEL, "| revision", REVISION)
import time
_t0 = _T_SESSION = time.time()   # the GPU session clock starts with the model load
llm = LLM(model=MODEL, revision=REVISION, tokenizer_revision=REVISION, quantization=SET["quantization"],
          max_model_len=MAX_LEN, gpu_memory_utilization=GPU_UTIL)
from s1a import log_v44 as L44
print("model load:", L44.save_load_time("results/wf_v50/model_load.json", time.time() - _t0, MODEL, REVISION))
```

```python
#@title 6. Backend: temperature 0, guided JSON per job schema, batches, cache per job (s1a/log_v44.VllmBackend44)
BATCH = 64  #@param {type:"integer"}
def sampling(max_tokens, schema):
    kw = dict(temperature=0, max_tokens=max_tokens)
    if schema is not None:
        try:
            from vllm.sampling_params import StructuredOutputsParams
            kw["structured_outputs"] = StructuredOutputsParams(json=schema)
        except ImportError:
            from vllm.sampling_params import GuidedDecodingParams
            kw["guided_decoding"] = GuidedDecodingParams(json=schema)
    return SamplingParams(**kw)
from s1a import log_v44 as L44
backend = L44.VllmBackend44(llm, MODEL, REVISION, sampling, max_len=MAX_LEN, batch=BATCH, vllm_version=vllm.__version__)
backend.model_id
```

```python
#@title 7. Data check: G4 jobs, laptop vectors, G3 columns and the loop-1 run files (git)
import sys
sys.path.insert(0, "/content/attribute-extraction/step1a/scripts")
import gpu_v50 as G
need = [G.G4 / d / f for d in G.SETS for f in ("manifest_names.json", "changed.json", "w2_vectors.npz", "prompts")]
need += [G.G3_FINAL / d / "final_columns.json" for d in G.SETS] + [G.LOOP1 / d / "loop2" for d in G.SETS]
need += [G.Y3 / d / f for d in G.SETS for f in ("entities.json", "a_prompts")]
missing = [str(p) for p in need if not p.exists()]
print("missing:", missing or "none")
assert not missing, "files missing: stop and tell Claude"
print({v: G.G48.config(v) for v in ("ARM",)})
```

```python
#@title 8. G4: rename the columns the context split changed (3 calls) -> results/wf_v50/g4/<set>/{responses/, renames.csv}
import subprocess
ren = G.g4(backend)
for dev, rows in ren.items():
    print(dev, [(r["entity"], r["cid"], r["old"], "->", r["new"]) for r in rows])
p = subprocess.run(["bash", "colab/sync.sh", "push-results", "wf_v50 G4 renames", "wf_v50"], capture_output=True, text=True)
print(p.stdout.strip() or p.stderr[-500:])
```

```python
#@title 9. Step 4 input build (CPU, laptop vectors) -> results/wf_v50/g4/w2/<set>/, results/wf_v50/g5/<set>/input/
import json
paths = G.build_inputs(G.CachedEnc())
for dev, p in paths.items():
    cols = json.loads(p.read_text(encoding="utf-8"))
    print(dev, p, len(cols), "columns")
```

```python
#@title 10. G5: Step 4 variant ARM, seeds 0-2, dev50 and dev50b -> results/wf_v50/g5/<set>/step4_ARM
MAX_CALLS = 140  #@param {type:"integer"}
import subprocess
from s1a import log_v44 as L44
res = G.g5(backend)
for (var, dev), rows in res.items():
    folder = G.G5 / dev / f"step4_{var}"
    calls = len(L44.call_rows(folder))
    kinds = {k: sum(x["repair_kinds"].get(k, 0) for x in rows) for k in ("p2_reask", "p2_reask2", "p2_reask_failed", "p2_reask2_failed")}
    print(var, dev, [(x["seed"], x["ok"], x["tables"], x["error"]) for x in rows], "| calls", calls, "| re-asks", kinds)
    assert calls <= MAX_CALLS, f"{var} {dev}: more calls than MAX_CALLS: stop and tell Claude"
p = subprocess.run(["bash", "colab/sync.sh", "push-results", "wf_v50 G5 Step 4 variant ARM", "wf_v50"], capture_output=True, text=True)
print(p.stdout.strip() or p.stderr[-500:])
```

```python
#@title 11. vLLM settings, GPU session time -> results/wf_v50/settings.json, then disconnect
import json, time, subprocess
def _seed():
    try:
        return llm.llm_engine.model_config.seed
    except Exception as e:
        return f"not readable: {type(e).__name__}"
info = {"chains": {f"{v} {d}": [(x["run"], x["ok"], x["tables"]) for x in rows] for (v, d), rows in res.items()},
        "vllm": vllm.__version__, "batch": BATCH, "batch_invariant_env": os.environ.get("VLLM_BATCH_INVARIANT"), "engine_seed": _seed(),
        "temperature": 0, "model": MODEL, "revision": REVISION, "reasks": {f"{d}": {k: sum(x["repair_kinds"].get(k, 0) for x in rows) for k in ("p2_reask", "p2_reask2")} for (v, d), rows in res.items()}, "renamed": {d: sum(r["renamed"] for r in rows) for d, rows in ren.items()}, "gpu_session_seconds": round(time.time() - _T_SESSION, 1),
        "model_load_seconds": L44.load_time("results/wf_v50/model_load.json")}
open("results/wf_v50/settings.json", "w").write(json.dumps(info, indent=1, default=str) + "\n")
print(json.dumps(info, indent=1, default=str))
p = subprocess.run(["bash", "colab/sync.sh", "push-results", "wf_v50 G4+G5 settings and session time", "wf_v50"], capture_output=True, text=True)
print(p.stdout.strip() or p.stderr[-500:])
print("Done: Runtime > Disconnect and delete runtime, then tell Claude 'GPU done'")
```
